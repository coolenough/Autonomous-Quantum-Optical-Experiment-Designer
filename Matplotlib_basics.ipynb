```python
left =fifa.loc[fifa['Preferred Foot']== 'Left'].count().iloc[0]
right =fifa.loc[fifa['Preferred Foot']== 'Right'].count().iloc[0]
colors=['#aabbff','#00ffff']
plt.pie([left,right],labels=['Left','Right'],colors=colors,autopct='%.2f%%') ##in a pie-chart give the all the list in a list of values first basically the list of numbers of different parameters
## labels are the lest of strings correspondingly
plt.title('Foot Preference of FIFA Players')


plt.show()
```

###Pie chart 2

```python
fifa.Weight
```

```python
fifa.Weight=[int(x.strip('lbs')) if type(x) ==str else x for x in fifa.Weight]
fifa.Weight
```

```python
fifa.Weight[0]####still string so to convert it into int type cast it in th else statement  only
```

```python
plt.style.use('ggplot')####edho colors marchadaaniki without typing all individually use avthundhi
light = fifa[fifa['Weight']<125].count().iloc[0]
light_medium=fifa[(fifa['Weight']>=125) & (fifa['Weight']<150)].count().iloc[0]
medium = fifa[(fifa['Weight']>=150) & (fifa['Weight']<175)].count().iloc[0]
medium_heavy=fifa[(fifa['Weight']>=175) & (fifa['Weight']<200)].count().iloc[0]
heavy = fifa[fifa['Weight']>=200].count().iloc[0]
weights = [light,light_medium,medium,medium_heavy,heavy]
explode = (0.4,0.1,0,0,0.4) ## by seeing the graph correspondingly you can notice the graph
##explode to break the pie  chart and by incresing the values in the tuple breaking distance increases
plt.pie(weights,labels=['Under 125','125-150','150-175','175-200','Over 200'],autopct='%.2f%%',pctdistance=0.8,explode=explode)##pctdistance antee pct marking for each undhi gaa adhhi ekkada display avvalani cheppadam like value lies between 0-1 ayithee markin is inside  the pie chart otherwise it will be outside
plt.title('Weight Distribution of FIFA Players (in lbs)',fontdict={'fontname':'cmr10'})
plt.show()

```

### Comparing Two teams using whiskers Chart




```python
fifa
```

```python
plt.style.use('default')
plt.figure(figsize=(6,11)) # Increased figure width slightly to give more space
barcelona =fifa.loc[fifa.Club == 'FC Barcelona']['Overall']
madrid =fifa.loc[fifa.Club == 'Real Madrid']['Overall']
revs = fifa.loc[fifa.Club == 'New England Revolution']['Overall']
boxes = plt.boxplot([barcelona,madrid,revs],tick_labels=['Barcelona',' Real Madrid','New England Revolution'],patch_artist=True,medianprops={'linewidth':2})

for box in boxes['boxes']:
  ##for setting the edge color.
  box.set(color='#00bbff',linewidth=3)
  ##Change fill color
  box.set(facecolor='#000000')
  ##face color pettali ante patch_artist=true ani pettali
plt.title("Professional  Soccer Team Comparision")
plt.xlabel("Teams")
plt.ylabel(" FIFA Overall Ratings")
plt.tight_layout() # Add tight_layout to prevent labels from overlapping
plt.show()
### so basically box and whiskers chart is for comparision of any no of teams based on some parameter here Overall points
##similar to piechart inputs first give list of series values of values and then give their list of strs labels
## orange line repredsents the median of the data

```
